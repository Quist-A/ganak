```python
import numpy as np
```

```python
def model_count(limdnnf):
    file = open(limdnnf, "r")
    content = file.readlines()
    file.close()

    # remove first line with metadata from limdnnf file
    content.pop(0)

    model_count = np.full(len(content),None) # model count per line

    for line in range(len(content)):
        if content[line][0] == "L":
            model_count[line] = 1  # this should be 2^{unassigned variables}
        elif content[line][0] == "A":
            model_count[line] = 1

            N_components = int(content[line].split()[1])
            components = []
            comparts = content[line].split()
            compart = 2
            for i in range(N_components):
                while ("[" in comparts[compart]) or ("]" in comparts[compart]) or (">" in comparts[compart]):
                    compart += 1
                components.append(comparts[compart])
                compart += 1
            for j in components:
                model_count[line] *= model_count[int(j)]
                
        elif content[line][0] == "O":  # assume deterministic DNNF
            if content[line].split()[1] == "0":
                model_count[line] = 0
            else:
                N_components = int(content[line].split()[2])
                components = []
                comparts = content[line].split()
                compart = 3
                for i in range(N_components):
                    while ("[" in comparts[compart]) or ("]" in comparts[compart]) or (">" in comparts[compart]):
                        compart += 1
                    components.append(comparts[compart])
                    compart += 1

                model_count[line] = 0
                for j in components:
                    model_count[line] += model_count[int(j)]
        else:
            print("could not handle line",line,"from input file")
                    
    #for i,mc in enumerate(model_count):
    #    print("line",i+1,":",mc)
    return(model_count[-1])
```
